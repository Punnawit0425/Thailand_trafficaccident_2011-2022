```python
print("Before:", df.shape)               # check before
df = df.drop_duplicates()                # remove duplicates
print("After:", df.shape)                # check after
print("Duplicates left:", df.duplicated().sum())  # should be 0
```

```python
df.isnull().sum()
```

```python
df["age"] = df["age"].fillna(df["age"].median())
df["age"].isnull().sum()
```

```python
df["age"].describe()
```

```python
df = df[(df["age"] >= 0) & (df["age"] <= 120)]
df["age"].describe()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()    # should now print 0
```

```python
df["gender"].value_counts()
```

```python
df["nationality"].value_counts()
```

```python
df["vehicle_type"].value_counts()
```

```python
df["vehicle_type"] = df["vehicle_type"].replace({"moterbike": "motorbike", "hevay_truck" : "heavy_truck",})
#Confirm the fix
df["vehicle_type"].value_counts()
```

```python
#Save clean data
df.to_csv("data/clean_data.csv", index=False)
```

```python
df["vehicle_type"].value_counts().plot(kind="bar")
plt.title("Accident by vehicle types")
plt.xlabel("Vehicle Type")
plt.ylabel("Number of accidents")
```

From the bar charts, we can say that the vehicle type with the most numbers of accident is unidentified. Among the identified vehicle, motorbike has the most accidents.

```python
df[df["vehicle_type"] != "unidentified"]["vehicle_type"].value_counts().plot(kind="bar")
plt.title("Accident by vehicle types")
plt.xlabel("Vehicle Type")
plt.ylabel("Number of accidents")
```

```python
df["gender"].value_counts().plot(kind="pie", autopct = "%1.1f%%")
plt.title("Gender pie chart")
plt.ylabel("")
plt.show()
```

```python
df["province_en"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 province")
plt.xlabel("Province")
plt.ylabel("Number of Accidents")
plt.show()
```

```python
df["age"].plot(kind="hist", bins = 30)
plt.title("Age Distributions of accident victims")
plt.xlabel("Age")
plt.ylabel("Number of accidents")
plt.show()
```

As the chart shows, number of accidents start increasing in the age of 15 due to more use of vehicle by that age.
Another important detail is that the spike char around 38 that goes up to 40,000 is due to putting median value, 38, in the missing ones, which causes this spike.

```python
import folium
from folium.plugins import HeatMap

# Sample 20,000 points so the map stays fast (still plenty for a heatmap)
sample = df[['latitude', 'longitude']].dropna().sample(20000, random_state=1)

# Center the map on Thailand
m = folium.Map(location=[13.7, 100.5], zoom_start=6)

# Add the heatmap layer
HeatMap(sample.values, radius=8).add_to(m)

# Save to a file you can open in your browser
m.save('accident_heatmap.html')
m
```
